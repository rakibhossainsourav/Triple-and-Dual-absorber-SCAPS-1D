```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
#1.1	Importing the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
#1.2	Importing the dataset
dataset = pd.read_csv('JSC.csv')

# Handle missing values by dropping rows with NaNs
dataset.dropna(inplace=True)

X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
#1.3	Splitting the dataset into the Training set and Test set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```
#1.4	Training the Regression models on the whole dataset
#For Linear regression
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
#For Support Vector regression
from sklearn.svm import SVR
regressor = SVR(kernel = 'rbf')
regressor.fit(X_train, y_train)
#For Decision Tree regression
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(random_state = 0)
regressor.fit(X_train, y_train)
#For Random Forest regression
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators = 10, random_state = 0)
regressor.fit(X_train, y_train)
#1.5	Predicting the Test Set Results
y_pred = regressor.predict(X_test)
np.set_printoptions(precision=2)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```
#1.6	Evaluating the model performance with cross validation
from sklearn.model_selection import KFold, cross_validate
from sklearn.metrics import make_scorer, mean_squared_error

def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))



# Define models
models = {
    "LR": LinearRegression(),
    "SVR": SVR(kernel='rbf'),
    "DTR": DecisionTreeRegressor(random_state=0),
    "RF": RandomForestRegressor(n_estimators=100, random_state=0)
}

# Cross-validation setup
kf = KFold(n_splits=5, shuffle=True, random_state=0)
scoring = {
    'R2': 'r2',
    'RMSE': make_scorer(rmse, greater_is_better=False)
}

# Storage for results
results = {}
for name, model in models.items():
    cv_results = cross_validate(model, X, y, cv=kf, scoring=scoring, return_train_score=False)
    r2_scores = cv_results['test_R2']
    rmse_scores = -cv_results['test_RMSE']  # negate since greater_is_better=False
    results[name] = {'R2': r2_scores, 'RMSE': rmse_scores}
    print(f"{name}: Mean R² = {np.mean(r2_scores):.4f} ± {np.std(r2_scores):.4f}, "
          f"Mean RMSE = {np.mean(rmse_scores):.4f} ± {np.std(rmse_scores):.4f}")

# Prepare boxplot data
r2_data = [results[m]['R2'] for m in models.keys()]
rmse_data = [results[m]['RMSE'] for m in models.keys()]

plt.figure(figsize=(10, 6))
# Two subplots: one for R², one for RMSE
plt.subplot(1, 2, 1)
plt.boxplot(
    r2_data, tick_labels=list(models.keys()), patch_artist=True,
    boxprops=dict(facecolor='skyblue', color='navy'),
    medianprops=dict(color='red', linewidth=2),
    whiskerprops=dict(color='navy', linewidth=1.5),
    capprops=dict(color='navy', linewidth=1.5)
)
plt.title('Cross-Validation R² Comparison', fontsize=13, fontweight='bold')
plt.ylabel('R² Score', fontsize=12, fontweight='bold')
plt.xticks(rotation=20, fontsize=11, fontweight='bold')
plt.yticks(fontsize=11, fontweight='bold')
plt.tick_params(direction='in', length=6, width=1.5)
plt.grid(alpha=0.3, linestyle='--')
plt.subplot(1, 2, 2)
plt.boxplot(
    rmse_data, tick_labels=list(models.keys()), patch_artist=True,
    boxprops=dict(facecolor='lightgreen', color='darkgreen'),
    medianprops=dict(color='red', linewidth=2),
    whiskerprops=dict(color='darkgreen', linewidth=1.5),
    capprops=dict(color='darkgreen', linewidth=1.5)
)
plt.title('Cross-Validation RMSE Comparison', fontsize=13, fontweight='bold')
plt.ylabel('RMSE', fontsize=12, fontweight='bold')
plt.xticks(rotation=20, fontsize=11, fontweight='bold')
plt.yticks(fontsize=11, fontweight='bold')
plt.tick_params(direction='in', length=6, width=1.5)
plt.grid(alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
import matplotlib.font_manager as fm
import matplotlib


matplotlib.rcParams['font.family'] = 'serif'
matplotlib.rcParams['font.serif'] = ['Times', 'DejaVu Serif', 'Georgia', 'serif']

# rest of your plotting code...


# ---- Customisation ----
label_fontsize = 16  # font size for labels and ticks
title_fontsize = 16  # font size for titles
legend_fontsize = 16  # font size for legend
save_dpi = 300
save_format = "png"

# Load dataset
dataset = pd.read_csv('JSC.csv')
dataset.dropna(inplace=True) # Added to handle missing values
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0)

# Train model
regressor = RandomForestRegressor(n_estimators=10, random_state=0)
regressor.fit(X_train, y_train)

# Predictions
y_train_pred = regressor.predict(X_train)
y_test_pred = regressor.predict(X_test)


def parity_plot(y_actual, y_pred, title, filename, color):
    plt.figure(figsize=(6, 6))
    plt.scatter(
        y_actual,
        y_pred,
        color=color,
        alpha=0.6,
        edgecolor='k',
        label='Predicted',
        linewidth=0.8
    )
    plt.plot(
        [8, 35],  # fixed x-range line
        [8, 35],  # fixed y-range line
        'r--',
        linewidth=2,
        label='Actual'
    )

    # Fix axis ranges
    plt.xlim(8, 35)
    plt.ylim(8, 35)

    # Labels & title with bold font and custom size
    plt.xlabel("Actual JSC",
               fontsize=label_fontsize,
               fontweight='bold')
    plt.ylabel("Predicted JSC",
               fontsize=label_fontsize,
               fontweight='bold')
    plt.title(title, fontsize=title_fontsize, fontweight='bold')

    # Legend without frame/outline, bold and larger font
    legend_font = fm.FontProperties(size=legend_fontsize, weight='bold')
    plt.legend(frameon=False, prop=legend_font)

    # Ticks: inward and bold tick labels with increased length and width
    plt.tick_params(direction='in',
                    length=10,
                    width=2,
                    colors='black',
                    labelsize=label_fontsize)
    plt.xticks(fontweight='bold')
    plt.yticks(fontweight='bold')

    # No grid
    plt.grid(False)

    plt.tight_layout()

    # Save plot
    plt.savefig(filename, dpi=save_dpi, format=save_format, bbox_inches='tight')
    plt.show()


# Generate and save plots
parity_plot(y_train, y_train_pred, "Training Data using Random Forest",
            "train_RF_plot.png", 'blue')
parity_plot(y_test, y_test_pred, "Testing Data using Random Forest",
            "test_RF_plot.png", 'green')
```

```
import shap
import pandas as pd

# Ensure X_test is a DataFrame with feature names
# Assuming 'feature_names' was defined earlier in the notebook
if not isinstance(X_test, pd.DataFrame):
    X_test_df = pd.DataFrame(X_test, columns=feature_names)
else:
    X_test_df = X_test

# Initialize the SHAP explainer with the trained regressor
# Assuming 'regressor' is the RandomForestRegressor trained previously
explainer = shap.Explainer(regressor)

# Calculate SHAP values for test set
shap_values = explainer.shap_values(X_test_df)

# Compute mean absolute SHAP value per feature to find top 8 features
mean_abs_shap = np.abs(shap_values).mean(axis=0)
top8_idx = np.argsort(mean_abs_shap)[::-1][:8]  # indices of top 8 features

# Select top 8 feature names and data
top8_features = X_test_df.columns[top8_idx]
X_test_top8 = X_test_df[top8_features]

# Select corresponding SHAP values for top 8 features
shap_values_top8 = shap_values[:, top8_idx]

# Plot SHAP summary beeswarm plot for top 8 features only (show=False to customize)
shap.summary_plot(shap_values_top8, X_test_top8, show=False)

# Customize plot: get current axes
ax = plt.gca()
# Customize x-axis ticks: increase font size, bold, inward ticks
ax.tick_params(axis='x', direction='in', length=10, width=2)
for label in ax.get_xticklabels():
    label.set_fontsize(16)
    label.set_fontweight('bold')

# Finally, show the plot
plt.show()
```
